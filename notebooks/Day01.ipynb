## 3. Python Virtual Environment

### What is a virtual environment?

A Python virtual environment keeps a project's Python packages isolated from other projects. This helps prevent dependency conflicts.

Bookify uses the virtual environment at the project root:

`Bookify/.venv/`

### Commands used

```powershell
python -m venv .venv
```

Creates a virtual environment named `.venv`.

```powershell
.\.venv\Scripts\python.exe --version
```

Checks the Python version associated with that environment.

### Problem encountered

PyCharm initially reported that no Python interpreter was configured. During troubleshooting, we discovered two virtual environments:

- `Bookify/.venv/`
- `Bookify/backend/.venv/`

FastAPI had initially been installed in the backend environment, while we wanted to standardise development on the root environment.

### Key lesson

- **Interpreter:** The Python executable used to run code.
- **Environment:** The interpreter and the packages installed alongside it.
- **Working directory:** The folder from which commands run. It affects relative paths and can affect Python imports.

We selected the root `.venv` as the standard development environment.

## 4. Building and Debugging the FastAPI Backend

### What we built

We created `backend/Main.py`, which contains our FastAPI application and two endpoints.

- `/` — Returns a welcome response.
- `/health` — Returns a simple health status to verify that the backend is responding.

### Important concepts

**Backend:** The part of an application that processes requests and handles application logic.

**API:** An interface that allows different parts of an application to communicate.

**Endpoint:** A URL path that performs a particular function when requested.

**FastAPI:** A Python framework used to build APIs.

### Starting the backend

We used Uvicorn to run the FastAPI application from the project root:

```powershell
python -m uvicorn backend.Main:app --reload
```

Breaking down the command:

- `python -m uvicorn` runs Uvicorn using the selected Python interpreter.
- `backend.Main` identifies the Python module containing our application.
- `:app` identifies the FastAPI instance inside that module.
- `--reload` restarts the server when relevant source files change during development.

### Debugging the startup error

Initially, Uvicorn reported that it could not find the expected `app` attribute in `backend.Main`.

We checked `backend/Main.py`, confirmed that the application was defined as `app = FastAPI(...)`, saved the file, and restarted the server.

**Key lesson:** In `module:variable`, the part before the colon identifies the module, while the part after it identifies the application object Uvicorn must load.

### Verification

We opened the endpoints in the browser and confirmed that they responded successfully:

- `/` returned the welcome response.
- `/health` returned `ok`.
- `/docs` opened FastAPI's interactive API documentation.

This confirmed that the initial backend was running and responding to requests.

## 5. Git and Version Control

### What is Git?

Git is a version control system that tracks changes to files in a project. It lets us review changes and return to earlier committed versions when necessary.

### Commands we used

```powershell
git init
```

Initialises a Git repository in the project folder.

```powershell
git status
```

Shows which files are untracked, modified, or staged.

```powershell
git add .
```

Stages changes so they can be included in the next commit.

```powershell
git diff --cached --stat
```

Summarises the staged changes.

```powershell
git commit -m "Initialise Bookify project"
```

Creates a snapshot of the staged changes.

```powershell
git log -1 --oneline
```

Displays the latest commit in a compact format.

### Our first commit

- **Commit ID:** `ee39f2d`
- **Message:** `Initialise Bookify project`

### Important distinction

1. Editing a file changes it on disk.
2. `git add` stages the changes for a future commit.
3. `git commit` saves a snapshot in the local Git history.
4. `git push` uploads committed history to a remote repository, such as GitHub.

A commit does not automatically upload anything to GitHub. Our local repository and remote repository are separate.

### Lesson learned

We edited the engineering journal after making our initial commit. Git then showed that the file had changed again.

This taught us that **a commit records the staged version of a file at that point in time**. Changes made afterwards need to be staged and committed separately.

### Current status

The local Git repository has been initialised and the first commit exists. Connecting the project to GitHub is a separate step.

## 6. Bookify Architecture

### What is Bookify?

Bookify is a personal knowledge discovery application built around books, playlists, and learning across different sources.

### Technology stack

| Component | Technology | Purpose |
|---|---|---|
| Backend | Python + FastAPI | Handles application logic and API requests |
| Frontend | React + Vite | Provides the user interface |
| Database | SQLite | Stores structured application metadata |
| File storage | Local filesystem | Stores uploaded PDF files |
| Version control | Git | Tracks changes to the project |

### Understanding how the components connect

The frontend is what the user interacts with. The backend processes requests from the frontend. The database stores structured information, while the filesystem stores the actual uploaded PDF files.

For example, when a user uploads a book:

1. The frontend sends the PDF to the backend.
2. The backend receives and validates the file.
3. The backend saves the PDF to local storage.
4. The backend extracts text from the PDF.
5. The application can store information about the book in SQLite.

This is our intended workflow, not a feature we've already completed.
